```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
```

## User-defined parameters

```python
# Problem parameters
c = 1          # Advection speed [m s-1]
dx = 500       # X resolution [m]
dt = 200       # Time step [s]
t0 = 0.        # Initial time
tmax = 0.1     # Duration of simulation [days]
xmax = 10.0e3  # Length of the basin [m]
```

## Grid definition

```python
# Time axis
NT = int(np.floor(tmax*86400/dt)) + 1  # number of steps, starts from 0

# X axis, centred at 0
x = np.arange(-xmax, xmax, dx)
IM = len(x)  # number of grid points

print('grid size:', IM)
print('number of timesteps:', NT)
```

## Spatial shape of the temperature disturbance

The initial temperature disturbance is a Gaussian peak.

```python
def Gaussian(x):
    # Gaussian peak function
    T0 = 15          # baseline temperature [degC]
    sigma = 1000.    # width of the peak [m]
    A = 5.           # amplitude [degC]
    y = T0 + A*np.exp(-(x/sigma)**2)
    return y
```

## Animation of the analytical solution

The analytical solution is displayed by shifting the Gaussian as:

$$T(x,t)=T(x-ct)$$

This means the temperature peak moves with speed \(c\), but keeps the same shape.

```python
fig, ax = plt.subplots()

# Set the axis limits and turn off autoscaling
plt.axis([-xmax/1000, xmax/1000, 13, 22])
plt.autoscale(False)

# Initial plot
line, = ax.plot(x/1000., Gaussian(x), 'k')
ax.set_xlabel('X (km)')
ax.set_ylabel('T ($^o$C)')
ax.set_title('Analytical advection solution')

def animate(i):
    t = t0 + i*dt
    line.set_ydata(Gaussian(x - c*t))  # update analytical solution
    return line,

# Create the animation
ani = animation.FuncAnimation(
    fig, animate, interval=200, blit=True, frames=NT, repeat=True
)

plt.close(fig)
HTML(ani.to_jshtml())
```

```python
def make_advection_animation(c=1, dx=500, sigma=1000, 
                             dt=200, t0=0, tmax=0.1, xmax=10.0e3,
                             T0=15, A=5, title='Analytical advection'):
    
    # time axis
    NT = int(np.floor(tmax*86400/dt)) + 1
    
    # space axis
    x = np.arange(-xmax, xmax, dx)
    
    # Gaussian temperature disturbance
    def Gaussian(x):
        return T0 + A*np.exp(-(x/sigma)**2)
    
    fig, ax = plt.subplots(figsize=(7, 4))
    
    ax.axis([-xmax/1000, xmax/1000, 13, 22])
    ax.autoscale(False)
    
    line, = ax.plot(x/1000, Gaussian(x), 'k')
    
    ax.set_xlabel('X (km)')
    ax.set_ylabel('T ($^o$C)')
    ax.set_title(title)
    
    def animate(i):
        t = t0 + i*dt
        line.set_ydata(Gaussian(x - c*t))
        return line,
    
    ani = animation.FuncAnimation(
        fig, animate, interval=200, blit=True, frames=NT, repeat=True
    )
    
    plt.close(fig)
    
    print('dx =', dx, 'm')
    print('c =', c, 'm/s')
    print('sigma =', sigma, 'm')
    
    return HTML(ani.to_jshtml())
```

```python
make_advection_animation(dx=1000, c=1, sigma=1000,
                         title='dx = 1000 m')
```

```python
make_advection_animation(dx=200, c=1, sigma=1000,
                         title='dx = 200 m')
```

```python
make_advection_animation(dx=500, c=0.5, sigma=1000,
                         title='c = 0.5 m/s')
```

```python
make_advection_animation(dx=500, c=2, sigma=1000,
                         title='c = 2 m/s')
```

```python
make_advection_animation(dx=500, c=1, sigma=500,
                         title='sigma = 500 m')
```

```python
make_advection_animation(dx=500, c=1, sigma=2000,
                         title='sigma = 2000 m')
```

1. Changing dx changes the grid resolution.
- a smaller dx gives more grid points and makes the peak smoother.
- a larger dx gives fewer grid points and makes the peak less resolved.

2. Changing c changes the speed of movement.
- A larger c moves the temperature peak faster.
- A smaller c moves the temperature peak slower.

3. Changing σ changes the width of the temperature peak.
- a smaller σ gives a narrow and steep peak.
- a larger σ gives a wider and smoother peak.
